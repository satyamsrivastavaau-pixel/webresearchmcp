```python
from transformers import pipeline
print("Loading local instruction-tuned model...")
chatbot = pipeline("text-generation", model="Qwen/Qwen2.5-0.5B-Instruct")
```

```python
# The single prompt contains the instruction, the examples (shots), and the new task
few_shot_prompt = """
Task: Task: Explain a computer science concept using a simple everyday analogy.

Concept: Array
Analogy: A row of numbered mailboxes at a post office. You can access any box instantly if you know its exact index number.

Concept: Stack
Analogy: A stack of dinner plates in a cafeteria. You can only add a new plate to the top, and you must remove the top plate before reaching the ones underneath.

Concept: Binary Search Tree
Analogy:"""

print("Running local inference engine...")

# We pass the raw text prompt directly to execute the few-shot execution
outputs = chatbot(few_shot_prompt, max_new_tokens=60, return_full_text=False)
```

```python
# Extract and clean the generated response
final_response = outputs[0]["generated_text"].strip()

print("\n--- Few-Shot Model Response ---")
print(f"Predicted Analogy: {final_response}")
```
